```python
from pqcrypto.kem.ml_kem_768 import generate_keypair, encrypt, decrypt
import hashlib

public_key, secret_key = generate_keypair()

ciphertext, shared_secret_1 = encrypt(public_key)
shared_secret_2 = decrypt(secret_key, ciphertext)

aes_key_1 = hashlib.sha256(shared_secret_1).digest()
aes_key_2 = hashlib.sha256(shared_secret_2).digest()

print("KEM OK:", shared_secret_1 == shared_secret_2)
print("AES key OK:", aes_key_1 == aes_key_2)

print("public_key:", len(public_key))
print("secret_key:", len(secret_key))
print("ciphertext:", len(ciphertext))
print("shared_secret:", len(shared_secret_1))
print("aes_key:", len(aes_key_1), aes_key_1.hex())
```
